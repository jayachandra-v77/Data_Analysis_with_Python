# **Exploratory Data Analysis**

# ***Gender***

```python
df.columns
```

```python
ax = sns.countplot(x ='Gender', data= df)
```

```python
ax = sns.countplot(x='Gender', data= df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```python
sales_gen = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender', y = 'Amount', data = sales_gen)
```

### From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men

## Age

```python
df.columns
```

```python
ax = sns.countplot(x = 'Age Group', data = df)
```

```python
ax = sns.countplot(x = 'Age Group', data = df, hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Total amount vs age group

sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending = False).head(10)
```

```python
sales_age
```

```python
sns.barplot(data = sales_age, x = 'Age Group', y = 'Amount')
```

### From above graphs we can see that most of the buyers are of age group between 26 - 35 female

## **State**

```python
# total number of orders from top 10 states

top_states = df.groupby(['State'], as_index = False)['Orders'].count().sort_values(by= 'Orders', ascending =False).head(10)
```

```python
top_states
```

```python

sns.set(rc={'figure.figsize':(15,5)})
ax = sns.barplot(data = top_states,x = 'State', y = 'Orders')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
## total amount/sales from top 10 states


top_sales_st = df.groupby(['State'],as_index= False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(10)
```

```python
top_sales_st
```

```python
sns.set(rc={'figure.figsize':(15,5)})
ax = sns.barplot(x = 'State', y = 'Amount', data = top_sales_st)
```

### From above graphs we can see that most of the orders are from uttar pradesh, maharastra and karnataka respectively

## Marital Status

```python
sns.set(rc={'figure.figsize':(5,5)})
ax = sns.countplot(data = df, x = 'Marital_Status')
for bars in ax.containers:
    ax.bar_label(bars)
```

## Occupation

```python
sns.set(rc= {'figure.figsize': (20,5)})
sns.countplot(data = df, x = 'Occupation')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_occupation = df.groupby(['Occupation'], as_index = False)['Amount'].count().sort_values(by = 'Amount', ascending = False).head(10)

sns.set(rc = {'figure.figsize': (20,5)})
yu = sns.barplot(data = sales_occupation, x = 'Occupation', y = 'Amount')

for bars in yu.containers:
    you.bar_label(bars)
```

### From above graphs we can see that most of the buyers are working in IT. Aviation and healthcare sector

## Product Category

```python
df.columns
```

```python
sns.set(rc= {'figure.figsize': (20,5)})
ax = sns.countplot(data = df, x = 'Product_Category')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_product = df.groupby(['Product_Category'], as_index = False)['Amount'].count().sort_values(by = 'Amount', ascending = False).head(10)

sns.set(rc = {'figure.figsize': (20,5)})
ax = sns.barplot(data = sales_product  , x = 'Product_Category', y = 'Amount')

for bars in ax.containers:
    ax.bar_label(bars)
```

## From above graphs we can see that most of the sold products are from Food, Cloting & Electronic category

```python
# top 10 most sold products (same thing as above)

fig1, axl = plt.subplots(figsize=(12,7))
df.groupby('Product_ID')['Orders'].sum().nlargest(10).sort_values(ascending=False).plot(kind='bar')
```
